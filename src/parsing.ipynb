```python
import pandas as pd
import numpy as np

def load_data(file_path):
    try:
        data = pd.read_csv(file_path)
        return data
    except Exception as e:
        print(f"Error loading data from {file_path}: {e}")
        return None

data = load_data('../data/Bank_Statement.csv')
```

```python
data = data.rename(columns={'Unnamed: 8': 'remark'})
data = data.rename(columns={'Unnamed: 7': 'WN'})
```

```python


def to_number(series):
    try:
        return pd.to_numeric(series, errors='coerce')
    except Exception as e:
        print(f"Error converting series to numeric: {e}")
        return series

def clean_tr(data):
    try:
        data['Transaction Reciepient']= data['Transaction Reciepient'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)
    except Exception as e:
        print(f"Error cleaning 'Transaction Reciepient' column: {e}")

def clean_remark(data):
    try:
        data["remark"] = data["remark"].where(~data["remark"].str.startswith("REF#", na=False), "")
        data["remark"] = data["remark"].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)
    except Exception as e:
        print(f"Error cleaning 'Remark' column: {e}")

        
def normalize_keys(data):
    try:
        data["Transaction Reciepient"] = data["Transaction Reciepient"].str.upper().str.strip()
        data["remark"] = data["remark"].str.upper().str.strip()
        return data
    except Exception as e:
        print(f"Error normalizing keys: {e}")


```

```python
normalize_keys(data)
clean_tr(data)
clean_remark(data)
data

```
